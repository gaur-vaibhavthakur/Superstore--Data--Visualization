```python
# Supersore_Cleaning.py
import pandas as pd
import numpy as np
df= pd.read_csv(r"C:\Users\user\Downloads\Superstore_ Data_ Visualization\Superstore.csv")
```

```python
#intital_insepection
print("iniital Shape: ",df.shape)
df.head()
```

```python
#checking_missing_values
df.isnull().sum()
```

```python
#convert_ Date_columns
df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce',dayfirst= True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce',dayfirst=True)
```

```python
# Drop rowswith missing critical data
df.dropna(subset=['Order Date', 'Ship Date', 'Sales', ], inplace=True)
```

```python
#  Create new columns
df['Order Month'] = df['Order Date'].dt.to_period('M').astype(str)
df['Days to Ship'] = (df['Ship Date'] - df['Order Date']).dt.days
```

```python
#Clean up text columns
df['Category'] = df['Category'].str.strip().str.title()
df['Sub-Category'] = df['Sub-Category'].str.strip().str.title()
df['Region'] = df['Region'].str.strip().str.title()
df['Segment'] = df['Segment'].str.strip().str.title()
```

```python
#Drop irrelevant columns 
if 'Postal Code' in df.columns:
    df.drop(columns=['Postal Code'], inplace=True)
```

```python
#  Save cleaned dataset
df.to_csv(r"C:\Users\user\Downloads\Superstore_ Data_ Visualization\Superstore_cleaned.csv", index=False)
```
